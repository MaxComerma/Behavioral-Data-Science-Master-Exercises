#### EJERCICIO 1

Cread una función que permita invertir la puntuación de los ítems de las escalas incorporadas en la base de datos *DATA\_coaches*. A pesar que en esta base de datos todas las escalas tienen el mismo rango en la escala ordinal (de 1 a 5 puntos) haced que la función permita trabajar con cualquier escala ordinal a la que se le quiera invertir la puntuación. (1.5 puntos)

```python
def invertir(valores, minimo, maximo):
    return abs(valores - maximo) + minimo
```

#### EJERCICIO 7

Cread una función que clasifique cada individuo de la muestra de acuerdo al siguiente criterio:
<font size=2>
\begin{cases}
\text{Si  } t3=t1: \text{  estable}\\
\text{Si  } t3<t1: \text{  inestable positivo (SCS y CBQ) o negativo (BPSSQ)}\\
\text{Si  } t3>t1: \text{  inestable negativo (SCS y CBQ) o positivo (BPSSQ)}\\
\end{cases}
</font>

Utilizad la nueva función para clasificar a los participantes en el estudio por separado por cada dimensión de interés (autocompasión, burnout y apoyo social percibido). Guardad estas clasificaciones como nuevas variables de la base de datos. ($1.5$ puntos)

```python
import numpy as np

def clasificar(t1, t3, dimension):
    clasi = np.where(np.isnan(t1) | np.isnan(t3), np.nan,
                     np.where(t3 == t1, "Estable",
                              np.where(t3 < t1,
                                       np.where(dimension in ["scs", "cbq"], "Inestable positivo", "Inestable negativo"),
                                       np.where(dimension in ["scs", "cbq"], "Inestable negativo", "Inestable positivo"))))
    return clasi

dfc["Clasificacion.SCS"] = clasificar(dfc["PT.t1scs"], dfc["PT.t3scs"], "scs")
dfc["Clasificacion.CBQ"] = clasificar(dfc["PT.t1cbq"], dfc["PT.t3cbq"], "cbq")
dfc["Clasificacion.SSQ"] = clasificar(dfc["PT.t1ssq"], dfc["PT.t3ssq"], "ssq")

dfc.replace("nan", np.nan, inplace=True)

print(dfc.iloc[:, 100:111])
```

#### EJERCICIO 8

¿Existe relación entre las variables cateǵoricas creadas en el apartado anterior? Realizad todos los análisis bivariantes posibles (escogiendo 2 variables cada vez). (1 punto)

```python
from scipy import stats

dfc8 = dfc.rename(columns={"Clasificacion.CBQ": "CBQ", "Clasificacion.SCS": "SCS", "Clasificacion.SSQ": "SSQ"})

contingenciaSCS_CBQ = pd.crosstab(dfc8["SCS"], dfc8["CBQ"], margins=True, margins_name="Total")
print(contingenciaSCS_CBQ)
print("\n")  

contingenciaSCS_SSQ = pd.crosstab(dfc8["SCS"], dfc8["SSQ"], margins=True, margins_name="Total")
print(contingenciaSCS_SSQ)
print("\n") 

contingenciaCBQ_SSQ = pd.crosstab(dfc8["CBQ"], dfc8["SSQ"], margins=True, margins_name="Total")
print(contingenciaCBQ_SSQ)
print("\n")  

chi2SCS_CBQ, pSCS_CBQ, _, _ = stats.chi2_contingency(observed = contingenciaSCS_CBQ.iloc[:-1, :-1])
chi2SCS_SSQ, pSCS_SSQ, _, _ = stats.chi2_contingency(observed = contingenciaSCS_SSQ.iloc[:-1, :-1])
chi2CBQ_SSQ, pCBQ_SSQ, _, _ = stats.chi2_contingency(observed = contingenciaCBQ_SSQ.iloc[:-1, :-1])

VcramerSCS_CBQ = np.sqrt(chi2SCS_CBQ / (contingenciaSCS_CBQ.iloc[:-1, :-1].sum().sum() * (min(contingenciaSCS_CBQ.iloc[:-1, :-1].shape)-1)))
VcramerSCS_SSQ = np.sqrt(chi2SCS_SSQ / (contingenciaCBQ_SSQ.iloc[:-1, :-1].sum().sum() * (min(contingenciaCBQ_SSQ.iloc[:-1, :-1].shape)-1)))
VcramerCBQ_SSQ = np.sqrt(chi2CBQ_SSQ / (contingenciaCBQ_SSQ.iloc[:-1, :-1].sum().sum() * (min(contingenciaCBQ_SSQ.iloc[:-1, :-1].shape)-1)))

print("SCS y CBQ:")
print("Ji-Cuadrado: " + str(round(chi2SCS_CBQ, 2)) + " + P-valor: " + str(round(pSCS_CBQ, 2)) + " + V de Cramér: " + str(round(VcramerSCS_CBQ, 2)))
print("\n")

print("SCS y SSQ:")
print("Ji-Cuadrado: " + str(round(chi2SCS_SSQ, 2)) + " + P-valor: " + str(round(pSCS_SSQ, 2)) + " + V de Cramér: " + str(round(VcramerSCS_SSQ, 2)))
print("\n")

print("CBQ y SSQ:")
print("Ji-Cuadrado: " + str(round(chi2CBQ_SSQ, 2)) + " + P-valor: " + str(round(pCBQ_SSQ, 2)) + " + V de Cramér: " + str(round(VcramerCBQ_SSQ, 2)))
print("\n")

print("Como se observa, no hay relación entre las variables")
```
